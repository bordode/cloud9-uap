## Key Assumptions
- UAP mass: 1000 kg (TicTac estimate)
- Antimatter storage efficiency: 50% (theoretical)
- Energy conversion efficiency: 90%
- Velocity range: Mach 1–10

## References

```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
c          = 3e8          # Speed of light (m/s)
uap_mass   = 1000         # kg — TicTac estimate
efficiency = 0.50 * 0.90  # storage efficiency × conversion efficiency

# Velocity range: Mach 1–10
velocity_mach = np.linspace(1, 10, 200)
velocity_ms   = velocity_mach * 343

# Kinetic energy and antimatter requirements
kinetic_energy = 0.5 * uap_mass * velocity_ms**2
antimatter_kg  = kinetic_energy / (efficiency * c**2)
antimatter_mg  = antimatter_kg * 1e6  # milligrams

# Plot
plt.figure(figsize=(11, 7))
plt.plot(velocity_mach, antimatter_mg, 'b-', linewidth=2, label='Antimatter Required (mg)')
plt.axhline(0.001, color='r', linestyle='--', linewidth=1.5, label='1 µg (Near-term tech)')
plt.axhline(1.0,   color='orange', linestyle='--', linewidth=1.5, label='1 mg (Advanced tech)')
plt.axvline(5.0,   color='green', linestyle=':', linewidth=1.5, label='Mach 5 (observed UAP)')
plt.xlabel('Velocity (Mach)', fontsize=13)
plt.ylabel('Antimatter Required (mg)', fontsize=13)
plt.title('Antimatter-Catalyzed Fusion: UAP Propulsion Requirements', fontsize=14)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.yscale('log')
plt.tight_layout()
plt.show()

# Key results table
print(f'{'Mach':>6} | {'KE (MJ)':>12} | {'Antimatter (mg)':>16}')
print('-' * 40)
for mach in [1, 3, 5, 7, 10]:
    idx = np.argmin(np.abs(velocity_mach - mach))
    ke_mj = kinetic_energy[idx] / 1e6
    print(f'{mach:>6} | {ke_mj:>12.2f} | {antimatter_mg[idx]:>16.6f}')
```

## Assembly Index (A_c) Connection

Antimatter-catalyzed fusion represents a **high-A_c propulsion system**:
- Requires multi-step assembly: antimatter production → storage → controlled annihilation → thrust
- Each step adds irreducible causal history → high assembly complexity
- Estimated A_c for full antimatter propulsion stack: **A_c > 50** (comparable to Perseus Cluster)
- Power density: ~90,000,000 MW/kg (theoretical maximum via E=mc²)
- Observed TicTac power density estimate: ~1.59 MW/kg

**Implication:** If UAP propulsion is antimatter-based, the observed power-to-weight ratio
is ~56 million times below the theoretical antimatter maximum — suggesting either:
1. Partial antimatter catalysis (small trigger → large fusion yield)
2. Inertial mass reduction (IMR) doing most of the work, antimatter just the ignition source
3. An entirely different mechanism we haven't modeled yet

# IMR + Antimatter Hybrid Engine Simulation
**Combining Inertial Mass Reduction with antimatter thrust for maximum performance.**

### Strategy:
1. Use IMR to reduce effective mass by 90%
2. Apply antimatter thrust to the reduced-mass craft
3. Result: 10x acceleration for same thrust budget

```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
c = 3e8
mu_0 = 4 * np.pi * 1e-7
g = 9.81

# Craft parameters
m_initial = 1000  # kg
imr_factors = np.linspace(0, 0.99, 100)  # 0% to 99% mass reduction
m_effective = m_initial * (1 - imr_factors)

# Antimatter thrust (fixed budget: 1 microgram antimatter)
antimatter_kg = 1e-9  # 1 microgram
thrust = antimatter_kg * c**2  # Joules -> effective thrust estimate

# Acceleration with IMR
acceleration = thrust / m_effective  # m/s²
acceleration_g = acceleration / g    # in Gs

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(imr_factors * 100, acceleration_g, 'b-')
plt.axhline(500, color='r', linestyle='--', label='TicTac observed (500 Gs)')
plt.xlabel('IMR Factor (%)')
plt.ylabel('Acceleration (Gs)')
plt.title('Acceleration vs. IMR Factor
(1 µg antimatter budget)')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(imr_factors * 100, m_effective, 'g-')
plt.xlabel('IMR Factor (%)')
plt.ylabel('Effective Mass (kg)')
plt.title('Effective Mass vs. IMR Factor')
plt.grid(True)

plt.tight_layout()
plt.show()

# Find IMR factor needed to achieve 500 Gs
target_g = 500
target_accel = target_g * g
required_mass = thrust / target_accel
required_imr = 1 - (required_mass / m_initial)
print(f'To achieve {target_g} Gs with 1µg antimatter:')
print(f'  Required effective mass: {required_mass:.2f} kg')
print(f'  Required IMR factor: {required_imr*100:.1f}%')
```
